```python
import pandas
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = pandas.read_csv(".\\Hades_II_Stats.csv")
print(data)
print(type(data))
```

```python
## Cumulative Win
cumulative_win = []
win_count = 0
for i in range(len(data["Win/Fail"])):
    if data["Win/Fail"][i] == "Win":
        win_count += 1
    cumulative_win.append(win_count)
data["Cumulative Win"] = cumulative_win
print(cumulative_win)
# print(len(cumulative_win))
# print(data["Night"])
# print(len(data["Night"]))
data.plot(x="Night", y="Cumulative Win")
```

```python
## Wins per weapon

weapons_list = ["Witch's Staff", "Sister Blades", "Moonstone Axe", "Argent Skull", "Umbral Flames", "Black Coat"]
wins_by_weapon = [0, 0, 0, 0, 0, 0]

for i in range(len(data["Win/Fail"])):
    if data["Win/Fail"][i] == "Win":
        weapon = data["Weapon"][i]
        if weapon == "Witch's Staff":
            wins_by_weapon[0] += 1
        if weapon == "Sister Blades":
            wins_by_weapon[1] += 1
        if weapon == "Moonstone Axe":
            wins_by_weapon[2] += 1
        if weapon == "Argent Skull":
            wins_by_weapon[3] += 1
        if weapon == "Umbral Flames":
            wins_by_weapon[4] += 1
        if weapon == "Black Coat":
            wins_by_weapon[5] += 1
            
print(wins_by_weapon)
total_wins = 0
for x in wins_by_weapon:
    total_wins += x
print(total_wins)

weapon_wins_comp = {
    'Weapon Uses': data["Weapon"].value_counts(),
    'Weapon Wins': wins_by_weapon
}


fig, ax = plt.subplots(layout='constrained')

x = np.arange(len(weapons_list))
bar_width = 0.25
multiplier = 0

for attribute, measurement in weapon_wins_comp.items():
    offset = bar_width * multiplier
    bars = ax.bar(x + offset, measurement, bar_width, label=attribute)
    ax.bar_label(bars, padding=3)
    multiplier += 1

ax.set_ylabel('Number')
ax.set_title('Weapons Wins')
ax.set_xticks(x + bar_width, weapons_list)
ax.legend(loc='upper left', ncols=3)
ax.set_ylim(0, 250)

plt.show()
```
